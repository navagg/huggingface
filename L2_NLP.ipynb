### Build the `chatbot` pipeline using Transformers Library

- Here is some code that suppresses warning messages.

````python
from transformers.utils import logging
logging.set_verbosity_error()
````

````python
from transformers import pipeline
````

- Define the conversation pipeline

````python
chatbot = pipeline(task="text-generation",
                   model="gpt2")
````

Info about ['blenderbot-400M-distill'](https://huggingface.co/facebook/blenderbot-400M-distill)

````python
user_message = """
What is the best learning activities to do in vacation?
"""
````

````python
# The 'Conversation' class from transformers.pipelines.conversational is not available or compatible with the 'text-generation' pipeline.
# We will manage conversation history manually using strings for the 'text-generation' pipeline.
````

````python
conversation_history = [user_message.strip()]
````

````python
print("Conversation history:\n" + "\n".join(conversation_history))
````

````python
formatted_prompt_parts = []
for i, message in enumerate(conversation_history):
    if i % 2 == 0: # User's turn
        formatted_prompt_parts.append(f"User: {message}")
    else: # Assistant's turn
        formatted_prompt_parts.append(f"Assistant: {message}")

# Add the final "Assistant:" to prompt GPT-2 for its response
input_text = "\n".join(formatted_prompt_parts) + "\nAssistant:"

response = chatbot(
    input_text,
    max_new_tokens=50, # Reduced to encourage conciseness
    return_full_text=False,
    stop_sequence='\nUser:', # Stop generation when a new user turn would begin
    repetition_penalty=1.2 # Add repetition penalty to reduce repetitive phrases
)
assistant_reply = response[0]['generated_text'].strip()

conversation_history.append(assistant_reply)
````

````python
print("Conversation history:\n" + "\n".join(conversation_history))
````

````python
new_user_message = "What else do you recommend?"
conversation_history.append(new_user_message.strip())
````

````python
print("Conversation history:\n" + "\n".join(conversation_history))
````
